```python
import random

print("----- CARD GAME -----")
try:
    players = int(input("Enter number of players: "))
    if players <= 0:
        print("Invalid number of players")
    elif 52 % players != 0:
        print("Number of players must divide 52 equally")
    else:
        suits = ["Hearts", "Diamonds", "Clubs", "Spades"]
        ranks = ["2", "3", "4", "5", "6", "7", "8", "9",
                 "10", "Jack", "Queen", "King", "Ace"]
        deck = [rank + " of " + suit for rank in ranks for suit in suits]
        random.shuffle(deck)
        hands = []
        for i in range(players):
            hands.append([])
        cards_each = 52 // players
        for i in range(52):
            player = i % players
            hands[player].append(deck[i])
        scores = [0 for i in range(players)]
        print()
        print("Cards distributed successfully!")
        print("Each player gets", cards_each, "cards")
        for round_number in range(1, cards_each + 1):
            print()
            print("----- Round", round_number, "-----")
            played_cards = []
            for i in range(players):
                card = random.choice(hands[i])
                hands[i].remove(card)
                played_cards.append(card)
                print("Player", i + 1, "played:", card)
            while True:
                try:
                    winner = int(input("Enter the round winner (player number): "))
                    if winner >= 1 and winner <= players:
                        scores[winner - 1] = scores[winner - 1] + 1
                        break
                    else:
                        print("Invalid player number")
                except ValueError:
                    print("Please enter a number")
        print()
        print("----- FINAL RESULT -----")
        for i in range(players):
            print("Player", i + 1, "won", scores[i], "rounds")
        highest_score = scores[0]
        for i in range(1, players):
            if scores[i] > highest_score:
                highest_score = scores[i]
        print()
        print("Highest score:", highest_score)
        winners = []
        for i in range(players):
            if scores[i] == highest_score:
                winners.append(i + 1)
        if len(winners) == 1:
            print("Overall Winner: Player", winners[0])
        else:
            print("It is a tie between:")
            for player in winners:
                print("Player", player)
except ValueError:
    print("Please enter a valid number")
```
